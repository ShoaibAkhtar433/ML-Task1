```python
blocked_words = ["suicide", "kill", "overdose", "self-harm"]
def safety_check(user_input):
  for word in blocked_words:
    if word in user_input.lower():
      return False
  return True
system_prompt = ""
```

```python
print("General Health Chatbot Started")

while True:
  query = input("You: ")

  if query.lower() == "exit":
    break

  if not safety_check(query):
    print("Bot: Please contact a healthcare professional or emergency services.")
    continue

  response = client.chat.completions.create(
      model = "gpt-3.5-turbo",
      messages = [
           {"role": "system", "content": system_prompt},
            {"role": "user", "content": query}
        ]
  )

  answer = response.choices[0].message.content
  print("Bot: ", answer)
```
